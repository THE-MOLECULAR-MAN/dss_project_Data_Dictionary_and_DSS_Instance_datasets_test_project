```python
import dataiku

from pprint import pprint
```

```python
# https://developer.dataiku.com/latest/api-reference/python/project-deployer.html#dataikuapi.dss.projectdeployer.DSSProjectDeployer
client = dataiku.api_client()
deployer = client.get_projectdeployer()
```

```python
deployer.list_stages()
```

```python
deployments = deployer.list_deployments(as_objects=True)
```

```python
for d in deployments:
    #     https://developer.dataiku.com/latest/api-reference/python/project-deployer.html#dataikuapi.dss.projectdeployer.DSSProjectDeployerDeployment
    #print(d.id)
    pprint(d.get_settings().get_raw())
```

```python
for i in deployer.list_infras():
    pprint(i.get_status().get_raw())
```
