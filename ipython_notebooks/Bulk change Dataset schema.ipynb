```python
import pandas as pd
from importlib.metadata import distributions
from typing import List, Dict, Any

def list_installed_packages() -> pd.DataFrame:
    """
    Retrieves a list of all installed Python packages and their versions 
    in the current environment and returns them as a Pandas DataFrame.
    """
    
    # Use importlib.metadata.distributions() for Python 3.8+
    installed_packages: List[Dict[str, Any]] = []
    
    for dist in distributions():
        try:
            installed_packages.append({
                "Name": dist.metadata["Name"],
                "Version": dist.version,
                "Summary": dist.metadata.get("Summary", "")
            })
        except Exception as e:
            # Handle potential errors with malformed distribution metadata
            installed_packages.append({
                "Name": str(dist),
                "Version": "Error",
                "Summary": f"Failed to retrieve metadata: {e}"
            })
            
    # Create the DataFrame and sort by package name
    df = pd.DataFrame(installed_packages)
    df = df.sort_values(by="Name").reset_index(drop=True)
    
    return df

# --- Execution ---

# 1. Get the list of installed packages as a DataFrame
installed_df = list_installed_packages()

# 2. Print to the Dataiku Log/Notebook Output (for inspection)
print("--- Installed Python Packages ---")
print(installed_df.to_string())

# --- Optional: If this is a Python Code Recipe (using Dataiku API) ---
# If you are writing a Python Code Recipe and want to save this list
# to an output dataset named 'installed_modules', use the following:

# import dataiku
# output_dataset = dataiku.Dataset("installed_modules") 
# output_dataset.write_with_schema(installed_df)
```
