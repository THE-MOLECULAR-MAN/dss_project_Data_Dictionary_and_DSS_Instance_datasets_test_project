```python
CONNECTION_TYPE='PostgreSQL'
OLD_SCHEMA_NAME='demo'
NEW_SCHEMA_NAME=''
```

```python
import dataiku
from dataiku import pandasutils as pdu
import pandas as pd
from json import pprint
```

```python
# Initialize the Dataiku API client
client = dataiku.api_client()

# Get the current project
project = client.get_default_project()

# List all datasets in the current project
dataset_names = set(dataset['name'] for dataset in project.list_datasets())
```

```python
for dataset_name in dataset_names:
    dataset_handle = dataiku.Dataset(dataset_name)
    dataset_info = dataset_handle.get_config()
    # print(dataset_name)

    # Check if the dataset is using the old S3 connection
    if dataset_info['type'] == CONNECTION_TYPE:
        if dataset_info['params']['schema'] == OLD_SCHEMA_NAME:
            settings = project.get_dataset(dataset_name).get_settings()
            print(settings.get_raw())
            break

            # Update to the new S3 connection
            #settings.set_schema()
            #settings.save()
            print(f"The schema for dataset {dataset_name} was changed from {OLD_SCHEMA_NAME} to {NEW_SCHEMA_NAME}")
```
